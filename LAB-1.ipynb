```python
import string
```

```python
exclude=string.punctuation
```

```python
def remove_punc1(text):
    exclude = set(string.punctuation)
    return text.translate(str.maketrans('', '', ''.join(exclude)))
```

```python
dataset['review'][10000]
```

```python
remove_punc1(dataset['review'][10000])
```

```python
dataset['review']=dataset['review'].apply(remove_punc1)
```

```python
import nltk
```

```python
from nltk.corpus import stopwords
stopwords.words('english')
```

```python
def remove_stopwords(text):
    stop_words = set(stopwords.words('english'))
    new_text = []

    for word in text.split():
        if word.lower() not in stop_words:
            new_text.append(word)

    return " ".join(new_text)
```

```python
remove_stopwords('I write entirely to find out what I\'m thinking, what I see and what it means. What I want and what I fear.')
```

```python
dataset['review'] = dataset['review'].apply(remove_stopwords)
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()

def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
stem_words('played, playfully, playing')
```

```python
dataset['review'] =dataset['review'].apply(stem_words)
```
