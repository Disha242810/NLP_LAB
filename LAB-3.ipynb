```python
import os
import re
import glob
import time
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import torch
import spacy
import transformers
from transformers import pipeline
from IPython.display import display

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

MODEL_NAME = "yangheng/deberta-v3-base-absa-v1.1"
MAX_REVIEWS = 5000
MIN_ASPECT_MENTIONS = 10
MAX_SENTENCES_PER_REVIEW = 8
OUTPUT_DIR = "/kaggle/working"
RANDOM_STATE = 42

os.makedirs(OUTPUT_DIR, exist_ok=True)

device = 0 if torch.cuda.is_available() else -1

print("Python environment ready")
print("Transformers:", transformers.__version__)
print("PyTorch:", torch.__version__)
print("Device:", torch.cuda.get_device_name(0) if device == 0 else "CPU")

if "reviews" not in globals() or not isinstance(reviews, pd.DataFrame):
    files = glob.glob("/kaggle/input/**/Reviews.csv", recursive=True)

    if not files:
        raise FileNotFoundError(
            "Reviews.csv not found. Attach the Amazon Fine Food Reviews dataset."
        )

    reviews = pd.read_csv(files[0])

reviews = reviews.copy()

required_columns = {"Text", "Score"}
if not required_columns.issubset(reviews.columns):
    raise ValueError("Dataset must contain Text and Score columns.")

reviews["Text"] = reviews["Text"].fillna("").astype(str).str.strip()
reviews["Score"] = pd.to_numeric(reviews["Score"], errors="coerce")
reviews = reviews.dropna(subset=["Score"])
reviews = reviews[reviews["Text"].str.len() > 0].copy()
reviews["Score"] = reviews["Score"].astype(int)

if "Summary" not in reviews.columns:
    reviews["Summary"] = ""

if "ProductId" not in reviews.columns:
    reviews["ProductId"] = "UNKNOWN_PRODUCT"

reviews["Summary"] = reviews["Summary"].fillna("").astype(str)
reviews["ProductId"] = reviews["ProductId"].fillna(
    "UNKNOWN_PRODUCT"
).astype(str)

def rating_to_sentiment(score):
    if score >= 4:
        return "positive"
    elif score <= 2:
        return "negative"
    return "neutral"

reviews["RatingSentiment"] = reviews["Score"].apply(rating_to_sentiment)
reviews = reviews.reset_index(drop=True)
reviews["ReviewRow"] = np.arange(len(reviews))

print("Dataset shape:", reviews.shape)
display(reviews.head())

# ============================================================
# 3. EXPLORATORY DATA ANALYSIS
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

sns.countplot(data=reviews, x="Score", ax=axes[0])
axes[0].set_title("Star Rating Distribution")

sns.countplot(
    data=reviews,
    x="RatingSentiment",
    order=["negative", "neutral", "positive"],
    ax=axes[1]
)
axes[1].set_title("Sentiment Derived from Ratings")

plt.tight_layout()
plt.show()

reviews["WordCount"] = reviews["Text"].str.split().str.len()

plt.figure(figsize=(9, 4))
sns.histplot(
    reviews["WordCount"].clip(
        upper=reviews["WordCount"].quantile(0.99)
    ),
    bins=40
)
plt.title("Review Length Distribution")
plt.xlabel("Word count")
plt.tight_layout()
plt.show()

# ============================================================
# 4. SELECT REVIEWS
# ============================================================

if MAX_REVIEWS is not None and len(reviews) > MAX_REVIEWS:
    work_reviews = reviews.sample(
        n=MAX_REVIEWS,
        random_state=RANDOM_STATE
    ).sort_values("ReviewRow").reset_index(drop=True)
else:
    work_reviews = reviews.copy()

print("Reviews selected:", len(work_reviews))

# ============================================================
# 5. LOAD SPACY POS TAGGER
# ============================================================

if "nlp" not in globals():
    nlp = spacy.load("en_core_web_sm")

# ============================================================
# 6. LOAD DEBERTA ABSA MODEL
# ============================================================

if "classifier" not in globals():
    classifier = pipeline(
        "text-classification",
        model=MODEL_NAME,
        device=device
    )

print("DeBERTa ABSA model ready.")

# ============================================================
# 7. POS TAGGING DEMONSTRATION
# ============================================================

example = (
    "The coffee tastes wonderful, "
    "but the plastic packaging is disappointing."
)

example_doc = nlp(example)

pos_preview = pd.DataFrame([
    {
        "Token": token.text,
        "Lemma": token.lemma_,
        "POS": token.pos_,
        "Detailed_Tag": token.tag_,
        "Dependency": token.dep_,
        "Head": token.head.text
    }
    for token in example_doc
])

display(pos_preview)

print("Noun phrases:")
for chunk in example_doc.noun_chunks:
    print("-", chunk.text)

# ============================================================
# 8. TEXT AND ASPECT EXTRACTION FUNCTIONS
# ============================================================

GENERIC_ASPECTS = {
    "i", "it", "this", "that", "they", "them", "we", "you",
    "product", "thing", "stuff", "time", "lot", "bit", "one",
    "something", "anything", "everything", "nothing",
    "amazon", "review", "purchase", "order", "item"
}

def normalize_text(text):
    return re.sub(r"\s+", " ", str(text)).strip()

def clean_aspect(aspect):
    aspect = aspect.lower().strip()
    aspect = re.sub(r"[^a-z0-9\s'-]", "", aspect)
    return re.sub(r"\s+", " ", aspect).strip(" -'")

def split_sentences(text):
    doc = nlp(normalize_text(text))
    return [
        sentence.text.strip()
        for sentence in doc.sents
        if sentence.text.strip()
    ]

def extract_aspects(sentence):
    """
    Extract noun phrases and individual nouns/proper nouns.
    These are aspect candidates, not guaranteed true product attributes.
    """
    doc = nlp(sentence)
    candidates = set()

    for chunk in doc.noun_chunks:
        candidate = clean_aspect(chunk.text)

        if (
            candidate
            and candidate not in GENERIC_ASPECTS
            and len(candidate.split()) <= 4
        ):
            candidates.add(candidate)

    for token in doc:
        if token.pos_ in {"NOUN", "PROPN"} and not token.is_stop:
            candidate = clean_aspect(token.lemma_)

            if (
                candidate
                and candidate not in GENERIC_ASPECTS
                and len(candidate) > 2
            ):
                candidates.add(candidate)

    return sorted(candidates)

# ============================================================
# 9. CREATE SENTENCE-ASPECT PAIRS
# ============================================================

pair_rows = []
review_metadata = {}

start_time = time.time()

for row in work_reviews.itertuples(index=False):
    review_id = int(row.ReviewRow)
    product_id = str(row.ProductId)
    score = int(row.Score)
    rating_sentiment = str(row.RatingSentiment)
    review_text = str(row.Text)

    review_metadata[review_id] = {
        "ProductId": product_id,
        "Score": score,
        "RatingSentiment": rating_sentiment,
        "ReviewText": review_text
    }

    sentences = split_sentences(review_text)
    sentences = sentences[:MAX_SENTENCES_PER_REVIEW]

    for sentence in sentences:
        for aspect in extract_aspects(sentence):
            pair_rows.append({
                "ReviewRow": review_id,
                "ProductId": product_id,
                "Score": score,
                "RatingSentiment": rating_sentiment,
                "Sentence": normalize_text(sentence),
                "Aspect": aspect
            })

pairs = pd.DataFrame(pair_rows)

if pairs.empty:
    raise RuntimeError("No aspect candidates were extracted.")

pairs = pairs.drop_duplicates(
    subset=["ReviewRow", "Sentence", "Aspect"]
).reset_index(drop=True)

print("Sentence-aspect pairs:", len(pairs))
print(
    "Unique sentence-aspect combinations:",
    len(pairs[["Sentence", "Aspect"]].drop_duplicates())
)
print("Extraction time:", round((time.time() - start_time) / 60, 2), "minutes")

display(pairs.head(10))

# ============================================================
# 10. RUN DEBERTA ABSA
# ============================================================

# Predict each unique sentence-aspect combination only once.
unique_pairs = pairs[["Sentence", "Aspect"]].drop_duplicates().reset_index(drop=True)

def normalize_model_label(label):
    label = str(label).lower().strip()

    if "positive" in label:
        return "positive"
    if "negative" in label:
        return "negative"
    if "neutral" in label:
        return "neutral"

    # For generic LABEL_n outputs, consult the model's label mapping.
    try:
        label_id = int(label.replace("label_", ""))
        id2label = classifier.model.config.id2label
        mapped_label = str(id2label[label_id]).lower()

        if "positive" in mapped_label:
            return "positive"
        if "negative" in mapped_label:
            return "negative"
        if "neutral" in mapped_label:
            return "neutral"
    except (ValueError, KeyError, TypeError, AttributeError):
        pass

    return label

def predict_aspect_sentiment(sentence, aspect):
    result = classifier(sentence, text_pair=aspect)

    if isinstance(result, list):
        result = result[0]

    raw_label = result.get("label", "unknown")

    return {
        "AspectSentiment": normalize_model_label(raw_label),
        "ConfidenceScore": float(result.get("score", np.nan)),
        "RawModelLabel": str(raw_label)
    }

prediction_cache = {}
prediction_errors = []

total_unique = len(unique_pairs)
start_time = time.time()

for i, row in enumerate(unique_pairs.itertuples(index=False), start=1):
    key = (
        row.Sentence.lower().strip()
        + " || "
        + row.Aspect.lower().strip()
    )

    try:
        prediction_cache[key] = predict_aspect_sentiment(
            row.Sentence,
            row.Aspect
        )
    except Exception as error:
        prediction_errors.append({
            "Sentence": row.Sentence,
            "Aspect": row.Aspect,
            "Error": str(error)
        })

    if i % 100 == 0 or i == total_unique:
        print(f"DeBERTa processed {i:,}/{total_unique:,}")

print(
    "Inference time:",
    round((time.time() - start_time) / 60, 2),
    "minutes"
)

if prediction_errors:
    print("Prediction errors:", len(prediction_errors))
    display(pd.DataFrame(prediction_errors).head())

# Attach predictions to every review-level aspect mention.
pairs["CacheKey"] = (
    pairs["Sentence"].str.lower().str.strip()
    + " || "
    + pairs["Aspect"].str.lower().str.strip()
)

pairs["AspectSentiment"] = pairs["CacheKey"].map(
    lambda key: prediction_cache.get(key, {}).get(
        "AspectSentiment", "error"
    )
)

pairs["ConfidenceScore"] = pairs["CacheKey"].map(
    lambda key: prediction_cache.get(key, {}).get(
        "ConfidenceScore", np.nan
    )
)

pairs["RawModelLabel"] = pairs["CacheKey"].map(
    lambda key: prediction_cache.get(key, {}).get(
        "RawModelLabel", "error"
    )
)

pairs["SentimentScore"] = pairs["AspectSentiment"].map({
    "positive": 1,
    "neutral": 0,
    "negative": -1
})

absa_df = pairs.copy()

print("Final prediction table:", absa_df.shape)
display(absa_df.head(15))

# ============================================================
# 11. EXPORT POS TAGGING RESULTS
# ============================================================

pos_rows = []

for sentence in pairs["Sentence"].drop_duplicates().head(500):
    doc = nlp(sentence)

    for token in doc:
        pos_rows.append({
            "Sentence": sentence,
            "Token": token.text,
            "Lemma": token.lemma_,
            "POS": token.pos_,
            "Detailed_Tag": token.tag_,
            "Dependency": token.dep_,
            "Head": token.head.text
        })

pos_df = pd.DataFrame(pos_rows)

pos_df.to_csv(
    os.path.join(OUTPUT_DIR, "pos_tagging_sample.csv"),
    index=False
)

print("POS tagging records:", len(pos_df))
display(pos_df.head(15))

# ============================================================
# 12. ASPECT FREQUENCY ANALYSIS
# ============================================================

aspect_frequency = (
    absa_df.groupby("Aspect")
    .agg(
        Mentions=("Aspect", "size"),
        UniqueReviews=("ReviewRow", "nunique"),
        UniqueProducts=("ProductId", "nunique")
    )
    .sort_values("Mentions", ascending=False)
    .reset_index()
)

print("Most frequently extracted aspects:")
display(aspect_frequency.head(25))

top_frequency = aspect_frequency.head(20).sort_values("Mentions")

plt.figure(figsize=(10, 7))
plt.barh(top_frequency["Aspect"], top_frequency["Mentions"])
plt.title("Most Frequently Mentioned Aspects")
plt.xlabel("Mentions")
plt.ylabel("Aspect")
plt.tight_layout()
plt.show()

# ============================================================
# 13. ASPECT SENTIMENT SUMMARY
# ============================================================

aspect_sentiment_summary = (
    absa_df.groupby(["Aspect", "AspectSentiment"])
    .size()
    .unstack(fill_value=0)
)

for label in ["negative", "neutral", "positive"]:
    if label not in aspect_sentiment_summary.columns:
        aspect_sentiment_summary[label] = 0

aspect_sentiment_summary = aspect_sentiment_summary[
    ["negative", "neutral", "positive"]
]

aspect_sentiment_summary["TotalMentions"] = (
    aspect_sentiment_summary[
        ["negative", "neutral", "positive"]
    ].sum(axis=1)
)

for label in ["negative", "neutral", "positive"]:
    aspect_sentiment_summary[label.title() + "Pct"] = (
        aspect_sentiment_summary[label]
        / aspect_sentiment_summary["TotalMentions"] * 100
    )

aspect_sentiment_summary["NetSentimentPct"] = (
    aspect_sentiment_summary["PositivePct"]
    - aspect_sentiment_summary["NegativePct"]
)

aspect_sentiment_summary = aspect_sentiment_summary.sort_values(
    "TotalMentions", ascending=False
)

display(aspect_sentiment_summary.head(25))

# ============================================================
# 14. ASPECT SENTIMENT VISUALIZATION
# ============================================================

chart_data = aspect_sentiment_summary.head(15).copy()

chart_pct = chart_data[
    ["negative", "neutral", "positive"]
].div(chart_data["TotalMentions"], axis=0) * 100

ax = chart_pct.sort_values("positive").plot(
    kind="barh",
    stacked=True,
    figsize=(11, 7)
)

ax.set_title("Sentiment Distribution by Aspect")
ax.set_xlabel("Percentage of mentions")
ax.set_ylabel("Aspect")
ax.legend(title="Sentiment", bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()
plt.show()

# ============================================================
# 15. MOST POSITIVE AND NEGATIVE ASPECTS
# ============================================================

eligible = aspect_sentiment_summary[
    aspect_sentiment_summary["TotalMentions"] >= MIN_ASPECT_MENTIONS
].copy()

best_aspects = eligible.sort_values(
    "NetSentimentPct", ascending=False
).head(15)

worst_aspects = eligible.sort_values(
    "NetSentimentPct", ascending=True
).head(15)

print("Most positive aspects:")
display(best_aspects[[
    "TotalMentions", "PositivePct", "NeutralPct",
    "NegativePct", "NetSentimentPct"
]])

print("Most negative aspects:")
display(worst_aspects[[
    "TotalMentions", "PositivePct", "NeutralPct",
    "NegativePct", "NetSentimentPct"
]])

# ============================================================
# 16. MODEL CONFIDENCE ANALYSIS
# ============================================================

confidence_summary = (
    absa_df.groupby("AspectSentiment")["ConfidenceScore"]
    .agg(["count", "mean", "median", "min", "max"])
    .round(4)
)

display(confidence_summary)

plt.figure(figsize=(9, 5))

sns.histplot(
    data=absa_df.dropna(subset=["ConfidenceScore"]),
    x="ConfidenceScore",
    hue="AspectSentiment",
    bins=25,
    element="step",
    stat="density",
    common_norm=False
)

plt.title("DeBERTa Confidence Score Distribution")
plt.tight_layout()
plt.show()

# ============================================================
# 17. PRODUCT-LEVEL ASPECT ANALYSIS
# ============================================================

product_aspect_summary = (
    absa_df.groupby(["ProductId", "Aspect", "AspectSentiment"])
    .size()
    .unstack(fill_value=0)
)

for label in ["negative", "neutral", "positive"]:
    if label not in product_aspect_summary.columns:
        product_aspect_summary[label] = 0

product_aspect_summary = product_aspect_summary[
    ["negative", "neutral", "positive"]
].reset_index()

product_aspect_summary["TotalMentions"] = (
    product_aspect_summary[
        ["negative", "neutral", "positive"]
    ].sum(axis=1)
)

product_aspect_summary["NetSentiment"] = (
    product_aspect_summary["positive"]
    - product_aspect_summary["negative"]
)

product_aspect_summary = product_aspect_summary[
    product_aspect_summary["TotalMentions"] >= 3
].sort_values(["ProductId", "Aspect"])

display(product_aspect_summary.head(20))

# ============================================================
# 18. COMPARE STAR RATINGS WITH ASPECT SENTIMENT
# ============================================================

rating_comparison = pd.crosstab(
    absa_df["RatingSentiment"],
    absa_df["AspectSentiment"],
    normalize="index"
).mul(100).round(2)

display(rating_comparison)

plt.figure(figsize=(8, 5))
sns.heatmap(rating_comparison, annot=True, fmt=".1f", cmap="Blues")
plt.title("Rating Sentiment vs Aspect Sentiment (%)")
plt.xlabel("DeBERTa aspect sentiment")
plt.ylabel("Rating-based sentiment")
plt.tight_layout()
plt.show()

# ============================================================
# 19. INSPECT SAMPLE PREDICTIONS
# ============================================================

for review_id in absa_df["ReviewRow"].drop_duplicates().head(5):
    metadata = review_metadata.get(review_id, {})

    print("\n" + "=" * 70)
    print("Review ID:", review_id)
    print("Product:", metadata.get("ProductId"))
    print("Star rating:", metadata.get("Score"))
    print("Rating sentiment:", metadata.get("RatingSentiment"))
    print("Review:", metadata.get("ReviewText", "")[:500])

    display(
        absa_df.loc[
            absa_df["ReviewRow"] == review_id,
            [
                "Sentence", "Aspect",
                "AspectSentiment", "ConfidenceScore"
            ]
        ].drop_duplicates()
    )

# ============================================================
# 20. PROJECT SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("PRODUCT REVIEW INTELLIGENCE SYSTEM")
print("=" * 60)
print("Reviews analyzed:", work_reviews["ReviewRow"].nunique())
print("Aspect-level records:", len(absa_df))
print("Unique aspect candidates:", absa_df["Aspect"].nunique())
print("Products represented:", absa_df["ProductId"].nunique())
print(
    "Average confidence:",
    round(absa_df["ConfidenceScore"].mean(), 4)
)

print("\nTop aspects:")
display(aspect_frequency.head(10))

print("\nStrongest positive aspects:")
display(best_aspects.head(10))

print("\nStrongest negative aspects:")
display(worst_aspects.head(10))

# ============================================================
# 21. EXPORT ALL RESULTS
# ============================================================

absa_df.to_csv(
    os.path.join(OUTPUT_DIR, "aspect_level_results.csv"),
    index=False
)

aspect_sentiment_summary.reset_index().to_csv(
    os.path.join(OUTPUT_DIR, "aspect_sentiment_summary.csv"),
    index=False
)

product_aspect_summary.to_csv(
    os.path.join(OUTPUT_DIR, "product_aspect_summary.csv"),
    index=False
)

aspect_frequency.to_csv(
    os.path.join(OUTPUT_DIR, "aspect_frequency.csv"),
    index=False
)

rating_comparison.to_csv(
    os.path.join(OUTPUT_DIR, "rating_vs_aspect_sentiment.csv")
)

print("\nExported files:")
for filename in [
    "aspect_level_results.csv",
    "aspect_sentiment_summary.csv",
    "product_aspect_summary.csv",
    "aspect_frequency.csv",
    "pos_tagging_sample.csv",
    "rating_vs_aspect_sentiment.csv"
]:
    path = os.path.join(OUTPUT_DIR, filename)
    if os.path.exists(path):
        print(filename, "-", round(os.path.getsize(path) / 1024, 1), "KB")

print("\nPipeline completed successfully.")
```
